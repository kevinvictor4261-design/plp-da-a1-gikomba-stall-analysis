# Wanjiku's Gikomba Stall Performance Analysis

## Task 1: Overall Sales Metrics & Extremes
**Approach:** We use built-in Python functions (`sum`, `len`, `max`, `min`, and `list.index()`) to compute total and average sales. Index positions are mapped to week numbers (indices 0–6 for Week 1, 7–13 for Week 2) and day names using the `days` list.

```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700]   # 14 days, KES
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2

stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions":   {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma":   {"qty": 9,  "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado":  {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}

total_sales = sum(daily_sales)
avg_sales = total_sales / len(daily_sales)

best_idx = daily_sales.index(max(daily_sales))
worst_idx = daily_sales.index(min(daily_sales))

best_day = days[best_idx]
best_week = (best_idx // 7) + 1
best_amount = daily_sales[best_idx]

worst_day = days[worst_idx]
worst_week = (worst_idx // 7) + 1
worst_amount = daily_sales[worst_idx]

print(f"Total Sales: KES {total_sales:,.2f}")
print(f"Average Daily Sales: KES {avg_sales:,.2f}")
print(f"Best Day: {best_day} (Week {best_week}) with KES {best_amount:,}")
print(f"Worst Day: {worst_day} (Week {worst_week}) with KES {worst_amount:,}")
```

## Task 2: Week-on-Week Growth Analysis
**Approach:** We define a function `week_totals(sales)` that splits the 14-day list into two 7-day slices, returns their sums as a tuple, and unpacks them to compute the percentage growth.

```python
def week_totals(sales):
    """
    Splits a 14-day sales list into two weekly totals using list slicing.
    Returns a tuple of (week_1_total, week_2_total).
    """
    week1 = sum(sales[:7])
    week2 = sum(sales[7:])
    return week1, week2

w1_total, w2_total = week_totals(daily_sales)
wow_growth = ((w2_total - w1_total) / w1_total) * 100

print(f"Week 1 Total: KES {w1_total:,}")
print(f"Week 2 Total: KES {w2_total:,}")
print(f"Week-on-Week Growth: {wow_growth:.2f}%")
```

## Task 3: Daily Performance Classification
**Approach:** We define `classify_day(amount, average)` to categorize daily earnings relative to the average threshold (>20% above, within ±20%, or >20% below). We then loop through all days and tally counts using a frequency dictionary.

```python
def classify_day(amount, average):
    """
    Classifies a sales amount as 'Strong', 'Normal', or 'Weak'
    based on a 20% variance threshold from the average.
    """
    if amount > average * 1.2:
        return "Strong"
    elif amount < average * 0.8:
        return "Weak"
    else:
        return "Normal"

classification_counts = {"Strong": 0, "Normal": 0, "Weak": 0}

for amount in daily_sales:
    status = classify_day(amount, avg_sales)
    classification_counts[status] += 1

print("Day Classifications Count:")
for status, count in classification_counts.items():
    print(f"- {status}: {count} days")
```

## Task 4: Weekday vs Weekend Performance Comparison
**Approach:** We separate weekday indices (Mon–Fri) from weekend indices (Sat–Sun) across both weeks, calculate their respective averages, and compute the percentage difference.

```python
weekday_indices = [0, 1, 2, 3, 4, 7, 8, 9, 10, 11]
weekend_indices = [5, 6, 12, 13]

weekday_sales = [daily_sales[i] for i in weekday_indices]
weekend_sales = [daily_sales[i] for i in weekend_indices]

avg_weekday = sum(weekday_sales) / len(weekday_sales)
avg_weekend = sum(weekend_sales) / len(weekend_sales)

diff_percentage = ((avg_weekend - avg_weekday) / avg_weekday) * 100

print(f"Average Weekday Sales: KES {avg_weekday:,.2f}")
print(f"Average Weekend Sales: KES {avg_weekend:,.2f}")
print(f"Weekend vs Weekday Difference: {diff_percentage:.2f}% higher on weekends")
```

## Task 5: Inventory Reorder Check & Gross Margin Calculation
**Approach:** We iterate through the stock dictionary using `.items()` to identify products with quantities at or below their reorder threshold and calculate individual gross margin percentages `(price - cost) / price`.

```python
def get_reorder_products(stock_dict):
    """
    Returns a list of product names whose current quantity
    is at or below their specified reorder level.
    """
    reorder_list = []
    for product, details in stock_dict.items():
        if details["qty"] <= details["reorder"]:
            reorder_list.append(product)
    return reorder_list

items_to_reorder = get_reorder_products(stock)

print("Products needing restock:", items_to_reorder)
print("\nGross Margins per Product:")
for product, details in stock.items():
    margin = ((details["price"] - details["cost"]) / details["price"]) * 100
    print(f"- {product}: {margin:.1f}%")
```

## Task 6: Stall Performance Report
**Approach:** We print a formatted summary using f-strings with thousands separators (`:,`) for all financial figures.

```python
print("=" * 45)
print("       STALL PERFORMANCE REPORT - GIKOMBA")
print("=" * 45)
print(f"Total 14-Day Revenue : KES {total_sales:,.2f}")
print(f"Daily Average Sales  : KES {avg_sales:,.2f}")
print(f"Week 1 Revenue       : KES {w1_total:,.2f}")
print(f"Week 2 Revenue       : KES {w2_total:,.2f}")
print(f"WoW Growth Rate      : {wow_growth:.2f}%")
print("-" * 45)
print(f"Top Performing Day   : {best_day} (KES {best_amount:,})")
print(f"Lowest Trading Day   : {worst_day} (KES {worst_amount:,})")
print(f"Items to Restock     : {', '.join(items_to_reorder)}")
print("=" * 45)
```
